# 2. Онтология ключевых терминов раздела

### Probability
- **тип:** mathematical foundation
- **определение:** численная мера правдоподобия события.
- **связи:** conditional probability, Bayes rule, language modeling.
- **роль:** базовый язык probabilistic NLP.

### Conditional Probability
- **тип:** mathematical relation
- **определение:** вероятность события $A$ при условии, что произошло событие $B$.
- **формула:**  
$$
P(A|B)=\frac{P(A \cap B)}{P(B)}
$$
- **связи:** Bayes rule, Naive Bayes, HMM, n-grams.

### Bayes Rule
- **тип:** probabilistic identity
- **определение:** выражает апостериорную вероятность через likelihood и prior.
- **формула:**  
$$
P(A|B)=\frac{P(B|A)P(A)}{P(B)}
$$
- **связи:** Naive Bayes, classification, autocorrect.

### Edit Distance
- **тип:** string similarity measure
- **определение:** минимальное число операций вставки, удаления, замены для превращения одной строки в другую.
- **связи:** autocorrect, dynamic programming.

### Language Model
- **тип:** probabilistic model
- **определение:** модель, присваивающая вероятность последовательности слов.
- **связи:** n-grams, perplexity, neural language models.

### N-gram
- **тип:** sequence representation / model component
- **определение:** последовательность из $n$ подряд идущих токенов.
- **связи:** Markov assumption, smoothing, perplexity.

### Smoothing
- **тип:** probability estimation technique
- **определение:** методы перераспределения вероятностной массы, чтобы не давать unseen событиям нулевую вероятность.
- **связи:** Laplace, Good-Turing, Kneser-Ney, backoff.

### Perplexity
- **тип:** evaluation metric
- **определение:** мера того, насколько хорошо language model предсказывает тестовый текст.
- **связи:** entropy, language modeling.

### Markov Chain
- **тип:** stochastic model
- **определение:** модель последовательности состояний, где следующее состояние зависит только от текущего.
- **связи:** HMM, transition probabilities.

### Hidden Markov Model
- **тип:** probabilistic sequence model
- **определение:** модель, в которой наблюдаемые слова порождаются скрытой последовательностью состояний.
- **связи:** POS tagging, Viterbi, emission/transition probabilities.

### Viterbi Algorithm
- **тип:** dynamic programming algorithm
- **определение:** алгоритм поиска наиболее вероятной последовательности скрытых состояний.
- **связи:** HMM, decoding.

# 3. Вероятности, условные вероятности и Bayes rule

## 3.1. Интуитивная идея

Probabilistic NLP начинается с очень простой идеи:

вместо того чтобы просто искать признаки, мы хотим **оценивать, насколько вероятен текст, слово, класс или последовательность состояний**.

Курс начинает этот блок именно с повторения probability и conditional probability, а затем выводит Bayes rule из определения условной вероятности. Bayes rule в курсе используется как общий фундамент для sentiment analysis, Naive Bayes и позже autocorrect.

## 3.2. Формальная постановка

### Обычная вероятность
$$
P(A)=\frac{\text{count}(A)}{\text{count}(\Omega)}
$$

### Условная вероятность
$$
P(A|B)=\frac{P(A \cap B)}{P(B)}
$$

### Bayes rule
$$
P(A|B)=\frac{P(B|A)P(A)}{P(B)}
$$

Где:

- $P(A)$ — **prior**;
- $P(B|A)$ — **likelihood**;
- $P(A|B)$ — **posterior**;
- $P(B)$ — **evidence**.

## 3.3. Пример

Пусть:

- $A$: твит позитивный;
- $B$: твит содержит слово *happy*.

Тогда:

$$
P(\text{positive}|\text{happy})
=
\frac{P(\text{happy}|\text{positive})P(\text{positive})}{P(\text{happy})}
$$

Именно такую постановку курс использует для объяснения, почему одно и то же слово может встречаться и в positive, и в negative corpus, но с разной вероятностью.

## 3.4. Исторический смысл

Вероятностный подход важен тем, что он переводит NLP из эвристик в математическую постановку:  
мы не просто говорим «это похоже на positive tweet», а оцениваем **вероятность класса**, **вероятность слова**, **вероятность последовательности**.

# 4. Edit Distance и Autocorrect

## 4.1. Интуитивная идея

Autocorrect можно описать как задачу:

**пользователь ввёл строку $x$, а система должна найти наиболее вероятное слово $w$, которое он имел в виду**.

Курс разбивает autocorrect на 4 шага:

1. обнаружить неправильное слово;
2. найти слова на расстоянии 1, 2, ... edits;
3. оставить только реальные слова;
4. выбрать наиболее вероятное исправление.

## 4.2. Вероятностная постановка

Нужно найти:

$$
\hat w = \arg\max_w P(w|x)
$$

По Bayes rule:

$$
\hat w = \arg\max_w P(x|w)P(w)
$$

Здесь:

- $P(w)$ — насколько слово вообще вероятно в языке;
- $P(x|w)$ — насколько вероятно получить опечатку $x$, если хотели написать $w$.

В базовой версии autocorrect часто упрощают до:

- сначала генерируют кандидатов по edit distance;
- потом ранжируют их по unigram probability $P(w)$.

## 4.3. Minimum Edit Distance

### Операции
- insertion;
- deletion;
- substitution;
- иногда transposition.

### Определение
Edit distance между строками $s$ и $t$ — минимальная стоимость последовательности операций, превращающей $s$ в $t$.

## 4.4. Dynamic Programming

Пусть $D[i,j]$ — минимальная стоимость преобразования первых $i$ символов строки $s$ в первые $j$ символов строки $t$.

Тогда:

$$
D[i,j] = \min
\begin{cases}
D[i-1,j] + \text{del}\\
D[i,j-1] + \text{ins}\\
D[i-1,j-1] + \text{sub}
\end{cases}
$$

где стоимость substitution равна 0, если символы совпадают.

## 4.5. Псевдокод

```text
function min_edit_distance(s, t):
    create matrix D of size (len(s)+1) x (len(t)+1)

    initialize first row and first column

    for i in 1..len(s):
        for j in 1..len(t):
            if s[i-1] == t[j-1]:
                cost = 0
            else:
                cost = 1

            D[i][j] = min(
                D[i-1][j] + 1,      # deletion
                D[i][j-1] + 1,      # insertion
                D[i-1][j-1] + cost  # substitution
            )

    return D[len(s)][len(t)]
```

## 4.6. Пример

Для `deah` → `dear` достаточно одной замены, поэтому distance = 1. Курс использует именно такой пример для объяснения autocorrect.

## 4.7. Применение в NLP

- spell checking;
- autocorrect;
- OCR post-processing;
- fuzzy search;
- string matching.

## 4.8. Ограничения

Edit distance учитывает **форму строки**, но не **контекст**.  
Поэтому `deer` и `dear` могут быть одинаково близки по символам, но правильный выбор зависит от фразы *happy birthday dear friend*. Курс отдельно подчёркивает, что контекстные ошибки — более сложная задача, не покрываемая базовым autocorrect.

## 4.9. Исторический контекст

До нейросетевых seq2seq spell check и autocorrect долгое время строились именно на комбинации:

**edit distance + frequency dictionary + probabilistic ranking**.

# 5. N-gram Language Models

## 5.1. Интуитивная идея

Language model присваивает вероятность целому предложению.  
Но считать полную вероятность последовательности слов напрямую слишком сложно:

$$
P(w_1,\dots,w_n)
$$

потому что каждое слово потенциально зависит от всей предыдущей истории.

Идея n-gram models: **приблизить эту зависимость коротким контекстом**.

## 5.2. Формальная постановка

По chain rule:

$$
P(w_1,w_2,\dots,w_n)=\prod_{i=1}^{n} P(w_i|w_1,\dots,w_{i-1})
$$

Это точно, но практически неудобно.

### Markov approximation

Для bigram model:

$$
P(w_i|w_1,\dots,w_{i-1}) \approx P(w_i|w_{i-1})
$$

Тогда:

$$
P(w_1,\dots,w_n)\approx \prod_{i=1}^{n} P(w_i|w_{i-1})
$$

Для trigram:

$$
P(w_i|w_1,\dots,w_{i-1}) \approx P(w_i|w_{i-2},w_{i-1})
$$

## 5.3. Оценка по частотам

### Unigram
$$
P(w)=\frac{\text{count}(w)}{\sum_{w'}\text{count}(w')}
$$

### Bigram
$$
P(w_i|w_{i-1})=\frac{\text{count}(w_{i-1},w_i)}{\text{count}(w_{i-1})}
$$

### Trigram
$$
P(w_i|w_{i-2},w_{i-1})
=
\frac{\text{count}(w_{i-2},w_{i-1},w_i)}
{\text{count}(w_{i-2},w_{i-1})}
$$

## 5.4. Алгоритм

1. preprocess corpus;
2. добавить специальные токены `<s>` и `</s>`;
3. посчитать n-gram counts;
4. оценить conditional probabilities;
5. для нового предложения перемножить вероятности n-gram.

## 5.5. Псевдокод

```text
function train_bigram_model(corpus):
    unigram_counts = {}
    bigram_counts = {}

    for sentence in corpus:
        tokens = ["<s>"] + sentence + ["</s>"]
        for i in range(len(tokens)):
            unigram_counts[tokens[i]] += 1
            if i > 0:
                bigram_counts[(tokens[i-1], tokens[i])] += 1

    return unigram_counts, bigram_counts
```

```text
function bigram_prob(w_prev, w_cur):
    return count(w_prev, w_cur) / count(w_prev)
```

## 5.6. Пример

Для фразы:

```text
I love NLP
```

bigram model вычисляет:

$$
P(I|<s>) \cdot P(love|I) \cdot P(NLP|love) \cdot P(</s>|NLP)
$$

Курс объясняет perplexity именно для bigram language models и показывает, что unigram, bigram и trigram дают тексты разного качества: unigram почти случайный, bigram уже лучше, trigram заметно естественнее.

## 5.7. Применение

- speech recognition;
- autocomplete;
- early machine translation;
- spelling correction;
- text generation baselines.

## 5.8. Ограничения

Главная проблема — **data sparsity**.

Даже если слова известны, нужный n-gram может ни разу не встретиться в корпусе. Тогда вероятность получается нулевой. Курс это разбирает на примере случая, когда слова известны, но сам bigram в корпусе отсутствует.

## 5.9. Исторический контекст

N-gram models были стандартом language modeling до нейросетевой эпохи. Даже сегодня они важны как:

- baseline;
- понятная вероятностная модель;
- историческая ступень к neural LMs.

Google Ngram Viewer выдает вероятности ngram:

* a teacher   679x10^(-6) % среди всех биграм
* the teacher 952x10^(-6) % среди всех биграм

а не условные вероятности P(teacher∣a).

# 6. Unknown Words и Vocabulary Truncation

## 6.1. Интуитивная идея

В реальном корпусе огромное число редких слов. Если включать все слова в vocabulary, модель станет слишком разреженной и нестабильной. Поэтому редкие слова часто заменяют на специальный токен:

$$
\text{UNK}
$$

Курс прямо обсуждает такой подход: слова, встречающиеся реже порога, заменяются на `UNK`. Но также предупреждает, что чрезмерное число `UNK` искусственно занижает perplexity и делает модель «якобы лучше», чем она есть на самом деле.

## 6.2. Ограничения

Если слишком агрессивно использовать `UNK`:

- vocabulary упрощается;
- perplexity становится deceptively low;
- модель начинает хорошо предсказывать `UNK`, а не реальный язык.

Это важное замечание: **метрика может улучшаться не потому, что модель лучше, а потому, что задача стала проще**.

# 7. Smoothing Techniques

## 7.1. Интуитивная идея

Если unseen n-gram получает вероятность 0, то вероятность целого предложения с этим n-gram тоже становится 0.

Это неприемлемо.

Поэтому нужна smoothing: мы немного уменьшаем вероятности seen событий и отдаём часть массы unseen событиям.

Курс на этом этапе подробно объясняет add-one / Laplace smoothing и саму мотивацию: без smoothing всё, чего не было в train corpus, считается невозможным.

## 7.2. Laplace Smoothing

Для bigram:

$$
P(w_i|w_{i-1})
=
\frac{\text{count}(w_{i-1}, w_i)+1}
{\text{count}(w_{i-1})+|V|}
$$

Для общего n-gram:

$$
P(w_n|w_{n-N+1}^{n-1})
=
\frac{\text{count}(w_{n-N+1}^{n})+1}
{\text{count}(w_{n-N+1}^{n-1})+|V|}
$$

## 7.3. Пример

Если bigram `John eats` не встречался, то без smoothing:

$$
P(\text{eats}|\text{John}) = 0
$$

С add-one smoothing он станет малым, но ненулевым. Это и есть центральный пример из курса.

## 7.4. Другие smoothing methods

Хотя курс в основном акцентирует add-one smoothing, в более полном учебном изложении нужно знать и более сильные методы.

### Good-Turing
Перераспределяет вероятность на основе частот частот.

### Kneser-Ney
Один из лучших классических методов smoothing для n-gram language models, особенно в higher-order models.

### Backoff
Если высокий порядок n-gram не найден, переходим к модели меньшего порядка.

### Interpolation
Комбинируем несколько порядков:

$$
P_{\text{interp}}(w_i|h)=
\lambda_3 P_{\text{tri}}+\lambda_2 P_{\text{bi}}+\lambda_1 P_{\text{uni}}
$$

## 7.5. Ограничения Laplace smoothing

Laplace smoothing легко понять, но она:

- слишком сильно перераспределяет массу;
- плохо масштабируется на больших vocabularies;
- обычно хуже Kneser-Ney в реальном LM.

Поэтому в индустриальных классических LM Laplace почти не используется как финальный метод, но как учебный — идеален.

# 8. Perplexity

## 8.1. Интуитивная идея

**Perplexity (перплексия)** — это метрика качества языковой модели, которая показывает, **насколько модель “удивляется” данным**.

Чем выше вероятность тестового текста под моделью, тем ниже perplexity. Курс это формулирует прямо: lower perplexity means the sentences are more likely to sound natural to human ears.

**Коротко**

Perplexity = насколько модель уверена в следующем токене<br>
= exp(cross-entropy)<br>
= “среднее количество вариантов выбора”<br>

## 8.2. Формальная постановка

Для тестовой последовательности $W = w_1,\dots,w_m$:

$$
\text{PP}(W)=P(w_1,\dots,w_m)^{-1/m}
$$

Эквивалентно:

$$
\text{PP}(W)=\sqrt[m]{\frac{1}{P(W)}}
$$

Для n-gram model:

$$
\text{PP}(W)=
\left(
\prod_{i=1}^{m} P(w_i|h_i)
\right)^{-1/m}
$$

В логарифмической форме:

$$
\log \text{PP}(W)
=
-\frac{1}{m}\sum_{i=1}^{m}\log P(w_i|h_i)
$$

## 8.3. Интерпретация

Перплексия отвечает на вопрос: “Насколько сложно модели предсказать следующий токен?”

- низкая perplexity → модель хорошо предсказывает тестовый корпус;
- высокая perplexity → модель плохо понимает распределение текста.

Курс приводит интуицию:  
если модель даёт тестовому набору высокую вероятность, perplexity будет близка к 1; если вероятность ничтожна, perplexity будет высокой. 

## 8.4. Связь с entropy

Perplexity тесно связана с entropy:

$$
\text{PP}(W)=2^{H(W)}
$$

в случае логарифмов по основанию 2.

Именно поэтому perplexity можно понимать как «эффективное число вариантов», между которыми модель колеблется.

## 8.5. Практические замечания

Курс отдельно предупреждает:

- сравнивать perplexity корректно только у моделей с одинаковым vocabulary;
- `UNK` может искусственно улучшать perplexity;
- логарифмическая форма часто удобнее вычислительно.

## 8.6. Ограничения perplexity

Perplexity полезна, но:

- не всегда хорошо коррелирует с downstream quality;
- не отражает factuality;
- не отражает stylistic quality;
- для современных LLM сама по себе уже недостаточна.

# 9. Markov Chains

## 9.1. Интуитивная идея

Markov chain моделирует последовательность состояний, где следующее состояние зависит только от текущего.

В NLP это естественно для задач, где структура последовательности важнее отдельных слов. В курсе Markov chains вводятся через POS tagging: если текущее слово — verb, следующий POS tag с большей вероятностью noun, чем verb.

## 9.2. Формальная постановка

Пусть $q_t$ — состояние в момент $t$. Тогда Markov property:

$$
P(q_t|q_1,\dots,q_{t-1}) = P(q_t|q_{t-1})
$$

Матрица переходов:

$$
A_{ij}=P(q_t=j|q_{t-1}=i)
$$

Сумма по строке:

$$
\sum_j A_{ij}=1
$$

Курс именно так объясняет transition probabilities и states как directed graph со взвешенными переходами.

## 9.3. Применение

- POS tagging;
- speech recognition;
- dialogue state modeling;
- event sequence modeling.

## 9.4. Ограничения

Обычная Markov chain моделирует только состояния, но не объясняет, как из них получаются наблюдаемые слова. Для этого нужен следующий шаг — HMM.

# 10. Hidden Markov Models (HMM)

## 10.1. Интуитивная идея

В реальном NLP мы часто видим **слова**, но не видим **скрытые лингвистические состояния**.

Например, в POS tagging:

- наблюдаемое: слово;
- скрытое: part-of-speech tag.

HMM моделирует именно это:

- скрытые состояния переходят друг в друга;
- каждое скрытое состояние порождает наблюдаемое слово.

Курс описывает HMM как модель с transition matrix и emission matrix, где hidden states — теги, а observables — слова корпуса.

## 10.2. Формальная постановка

Пусть:

- $q_t$ — скрытое состояние;
- $x_t$ — наблюдаемое слово.

### Параметры модели
- начальные вероятности:
$$
\pi_i = P(q_1=i)
$$

- transition probabilities:
$$
A_{ij}=P(q_t=j|q_{t-1}=i)
$$

- emission probabilities:
$$
B_{j}(x)=P(x_t=x|q_t=j)
$$

## 10.3. Вероятность совместной последовательности

$$
P(x_{1:T}, q_{1:T})
=
P(q_1)\prod_{t=2}^{T}P(q_t|q_{t-1})
\prod_{t=1}^{T}P(x_t|q_t)
$$

Это ключевая factorization HMM.

## 10.4. Обучение параметров

### Transition matrix
$$
A_{ij}=
\frac{\text{count}(q_{t-1}=i, q_t=j)}
{\text{count}(q_{t-1}=i)}
$$

### Emission matrix
$$
B_j(x)=
\frac{\text{count}(q_t=j, x_t=x)}
{\text{count}(q_t=j)}
$$

Курс показывает именно такой counting-based способ заполнения transition и emission matrices из размеченного корпуса.

## 10.5. Сглаживание

Как и в n-gram models, transition и emission probabilities тоже могут страдать от нулевых значений. Курс отдельно упоминает smoothing для transition and emission matrices ради better generalization.

## 10.6. Пример

Для предложения:

```text
I love to learn
```

HMM ищет наиболее вероятную последовательность POS tags, учитывая:

- какие теги обычно следуют друг за другом;
- какие слова обычно испускаются каждым тегом.

Курс разбирает игрушечный пример, где `love` может быть и noun, и verb, а выбор зависит от transition × emission probabilities.

## 10.7. Применение

- POS tagging;
- named entity recognition в ранних системах;
- speech recognition;
- shallow parsing.

## 10.8. Ограничения

- строгая Markov assumption;
- emission зависит только от текущего hidden state;
- плохо учитывает дальний контекст;
- признаки почти не гибкие;
- трудно моделировать богатую зависимость между словами.

Именно из этих ограничений позже появились CRF, RNN, BiLSTM-CRF и transformers.

# 11. Viterbi Algorithm

Алгоритм Витерби — это алгоритм динамического программирования, который находит наиболее вероятную последовательность скрытых состояний по наблюдаемой последовательности. Чаще всего его используют вместе со скрытыми марковскими моделями (HMM).

## 11.1. Интуитивная идея

В HMM нужно не просто оценить вероятность, а найти **самую вероятную последовательность скрытых состояний** для данного предложения.

Перебирать все возможные tag sequences слишком дорого.

Viterbi решает это через dynamic programming.

Курс прямо подчёркивает, что хотя формулы выглядят тяжело, по сути Viterbi — это graph algorithm для поиска наиболее вероятного пути по скрытым состояниям.

## 11.2. Формальная постановка

Нужно найти:

$$
\hat q_{1:T}=\arg\max_{q_{1:T}} P(q_{1:T},x_{1:T})
$$

Определим:

$$
V_t(j)=\max_{q_{1:t-1}} P(q_{1:t-1}, q_t=j, x_{1:t})
$$

Тогда рекуррентное соотношение:

$$
V_t(j)=
\max_i \left[V_{t-1}(i)A_{ij}\right]B_j(x_t)
$$

И backpointer:

$$
D_t(j)=\arg\max_i \left[V_{t-1}(i)A_{ij}\right]
$$

## 11.3. Алгоритм

1. Инициализация для первого слова.
2. Для каждого следующего слова:
   - для каждого состояния $j$:
     - найти лучшее предыдущее состояние $i$;
     - сохранить score;
     - сохранить backpointer.
3. На последнем шаге взять state с максимальным score.
4. Восстановить путь назад по backpointers.

## 11.4. Псевдокод

```text
function viterbi(observations, A, B, pi):
    initialize V[1][j] = pi[j] * B[j][x1]
    initialize backpointer[1][j] = 0

    for t in 2..T:
        for each state j:
            V[t][j] = max_i(V[t-1][i] * A[i][j] * B[j][xt])
            backpointer[t][j] = argmax_i(V[t-1][i] * A[i][j])

    best_last_state = argmax_j V[T][j]
    backtrack using backpointer
    return best_path
```

## 11.5. Логарифмическая форма

Чтобы избежать numerical underflow:

$$
\log V_t(j)=
\max_i \left[\log V_{t-1}(i)+\log A_{ij}+\log B_j(x_t)\right]
$$

Курс специально предупреждает, что при умножении большого числа малых вероятностей надо переходить к log probabilities.

## 11.6. Применение

- POS tagging;
- speech recognition;
- decoding in sequence models;
- ранние bioinformatics задачи.

## 11.7. Исторический смысл

Viterbi — один из ключевых алгоритмов probabilistic sequence modeling. Даже если позже ты работаешь с transformers, понимание Viterbi важно как образец **динамического программирования для декодирования последовательностей**.

# 12. Связь probabilistic models с Naive Bayes

Хотя Naive Bayes уже упоминался в первом разделе, по сути он тоже probabilistic model.

Курс даёт для него полный probabilistic вывод:

- считаются priors;
- оцениваются conditional probabilities слов по классам;
- затем для документа считается product of likelihood ratios;
- чтобы избежать underflow, переходят к log-likelihood;
- решающее правило становится порогом по 0 в log-space.

## 12.1. Формула Naive Bayes

$$
P(c|d)\propto P(c)\prod_{i=1}^{n}P(w_i|c)
$$

В log-space:

$$
\log P(c|d)=\log P(c)+\sum_i \log P(w_i|c) + \text{const}
$$

Или для binary classification через ratio:

$$
\log \frac{P(c_+|d)}{P(c_-|d)}
=
\log \frac{P(c_+)}{P(c_-)}
+
\sum_i \log \frac{P(w_i|c_+)}{P(w_i|c_-)}
$$

Курс обозначает логарифм отношения условных вероятностей слова как $\lambda$, а log-likelihood документа получает суммой этих $\lambda$. 

## 12.2. Почему это важно именно здесь

Naive Bayes удобно понимать не как изолированный классификатор, а как первый реально полезный пример того, как probabilistic reasoning применяется к тексту.

Он показывает:

- что prior matters;
- что likelihood matters;
- что zero probabilities ломают модель;
- что log-domain вычисления важны;
- что independence assumption упрощает жизнь, но и ограничивает модель.

# 13. Типичные ошибки probabilistic NLP

Курс делает сильный акцент не только на формулах, но и на error analysis. Это важно.

## 13.1. Потеря смысла на preprocessing
Если удалить `not`, меняется полярность фразы.  
Если удалить sad-face punctuation, можно уничтожить негативный сигнал.

## 13.2. Independence assumption
Naive Bayes предполагает независимость слов, хотя в реальном языке слова зависимы. Курс приводит пример `sunny` и `hot`, которые часто встречаются вместе.

## 13.3. Word order
Фразы:

- *I am happy because I did not go*
- *I am not happy because I did not go*

могут содержать почти один и тот же bag of words, но иметь разный смысл. Курс прямо выделяет word order как источник ошибок.

## 13.4. Sarcasm / irony / euphemism
Чисто probabilistic bag-of-words методы плохо работают с сарказмом. Курс приводит пример позитивного movie review, который после preprocessing выглядит набором “негативных” слов.

## 13.5. Data sparsity
Редкие слова и unseen n-grams дают нули.  
Отсюда потребность в smoothing.

## 13.6. Numerical underflow
При перемножении большого числа вероятностей получаются очень маленькие числа.  
Отсюда обязательный переход к log-probabilities и в Naive Bayes, и в Viterbi.

# 14. Практический pipeline probabilistic NLP

```text
raw text
→ preprocessing
→ counts / frequency tables
→ probability estimation
→ smoothing
→ inference / decoding
→ evaluation
```

### Для autocorrect
```text
typed word
→ candidate generation by edit distance
→ filter valid words
→ rank by probability
→ choose best correction
```

### Для n-gram LM
```text
corpus
→ n-gram counts
→ conditional probabilities
→ smoothing
→ sentence probability
→ perplexity
```

### Для HMM tagging
```text
tagged corpus
→ transition matrix
→ emission matrix
→ smoothing
→ Viterbi decoding
→ best tag sequence
```

# 15. Короткие Python-примеры

## 15.1. Minimum Edit Distance

```python
def min_edit_distance(s: str, t: str) -> int:
    m, n = len(s), len(t)
    dp = [[0] * (n + 1) for _ in range(m + 1)]

    for i in range(m + 1):
        dp[i][0] = i
    for j in range(n + 1):
        dp[0][j] = j

    for i in range(1, m + 1):
        for j in range(1, n + 1):
            cost = 0 if s[i - 1] == t[j - 1] else 1
            dp[i][j] = min(
                dp[i - 1][j] + 1,      # delete
                dp[i][j - 1] + 1,      # insert
                dp[i - 1][j - 1] + cost
            )
    return dp[m][n]

print(min_edit_distance("deah", "dear"))  # 1
```

## 15.2. Bigram counts

```python
from collections import Counter

def bigram_counts(corpus):
    uni = Counter()
    bi = Counter()

    for sent in corpus:
        tokens = ["<s>"] + sent.split() + ["</s>"]
        uni.update(tokens)
        bi.update(zip(tokens[:-1], tokens[1:]))

    return uni, bi

corpus = [
    "i love nlp",
    "i love python",
    "nlp loves data"
]

uni, bi = bigram_counts(corpus)
print(bi[("i", "love")])
```

## 15.3. Bigram probability with Laplace smoothing

```python
def bigram_prob(prev_word, word, uni, bi, vocab_size):
    return (bi[(prev_word, word)] + 1) / (uni[prev_word] + vocab_size)
```

## 15.4. Sentence probability

```python
def sentence_prob(sentence, uni, bi, vocab):
    tokens = ["<s>"] + sentence.split() + ["</s>"]
    p = 1.0
    V = len(vocab)
    for a, b in zip(tokens[:-1], tokens[1:]):
        p *= bigram_prob(a, b, uni, bi, V)
    return p
```

## 15.5. Perplexity

```python
import math

def perplexity(sentence, uni, bi, vocab):
    tokens = ["<s>"] + sentence.split() + ["</s>"]
    V = len(vocab)
    log_prob = 0.0
    m = len(tokens) - 1

    for a, b in zip(tokens[:-1], tokens[1:]):
        p = bigram_prob(a, b, uni, bi, V)
        log_prob += math.log(p)

    return math.exp(-log_prob / m)
```

## 15.6. Простая HMM decoding заготовка

```python
def viterbi(obs, states, start_p, trans_p, emit_p):
    V = [{}]
    back = {}

    for st in states:
        V[0][st] = start_p[st] * emit_p[st].get(obs[0], 1e-12)
        back[st] = [st]

    for t in range(1, len(obs)):
        V.append({})
        new_back = {}

        for st in states:
            prob, prev_st = max(
                (
                    V[t - 1][pst] * trans_p[pst].get(st, 1e-12) * emit_p[st].get(obs[t], 1e-12),
                    pst
                )
                for pst in states
            )
            V[t][st] = prob
            new_back[st] = back[prev_st] + [st]

        back = new_back

    prob, state = max((V[-1][st], st) for st in states)
    return prob, back[state]
```

# 17. Итог раздела

Probabilistic NLP — это этап, на котором обработка языка стала по-настоящему математической.

Здесь появляются три большие идеи:

### 1. Язык как распределение вероятностей
Через Bayes rule, conditional probabilities и language modeling.

### 2. Последовательность как стохастический процесс
Через Markov chains, n-grams и HMM.

### 3. Инференс как алгоритмическая задача
Через dynamic programming, edit distance и Viterbi.

При этом именно здесь становятся очевидны и ограничения классического probabilistic NLP:

- короткий контекст;
- independence assumptions;
- плохая работа с дальними зависимостями;
- сильная зависимость от sparsity;
- ограниченная выразительность.

Из этих ограничений естественно вырастает следующий этап:

**probabilistic sequence models → neural sequence models → RNN / LSTM / GRU**.
